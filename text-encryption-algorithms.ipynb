```python
import re

def text_to_bits(text):
    """
    Преобразование текста в бинарный формат.
    """
    bits = []
    for char in text:
        bin_char = bin(ord(char))[2:].zfill(8)
        bits.extend([int(bit) for bit in bin_char])
    return bits

def bits_to_text(bits):
    """
    Преобразование бинарных данных в текст.
    """
    text = ""
    for i in range(0, len(bits), 8):
        byte = bits[i:i+8]
        char = chr(int("".join([str(bit) for bit in byte]), 2))
        text += char
    return text

def xor_bits(a, b):
    """
    XOR для бинарных данных.
    """
    return [int(bool(x) ^ bool(y)) for x, y in zip(a, b)]

def encode(message, key):
    """
    Кодирование сообщения на основе ключа.
    """
    message_bits = text_to_bits(message)
    key_bits = text_to_bits(key)

    # Приводим ключ к длине сообщения
    while len(key_bits) < len(message_bits):
        key_bits *= 2

    # Выполняем XOR для каждого бита сообщения и ключа
    encoded_bits = xor_bits(message_bits, key_bits)

    return encoded_bits

def decode(encoded_bits, key):
    """
    Декодирование сообщения на основе ключа.
    """
    key_bits = text_to_bits(key)

    # Приводим ключ к длине сообщения
    while len(key_bits) < len(encoded_bits):
        key_bits *= 2

    # Выполняем XOR для каждого бита закодированного сообщения и ключа
    message_bits = xor_bits(encoded_bits, key_bits)

    return bits_to_text(message_bits)

# Пример использования
message = "Фбвгдеёзж.,"
key = "ключ"

# Кодирование сообщения
encoded_bits = encode(message, key)
encoded_text = bits_to_text(encoded_bits)
print("Закодированное сообщение:", encoded_text)

# Декодирование сообщения
decoded_text = decode(encoded_bits, key)
print("Раскодированное сообщение:", decoded_text)
```
